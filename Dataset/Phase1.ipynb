## 1. Project Goal
The primary objective of this project is to apply data mining techniques to the **Airline Passenger Satisfaction** dataset. Our task involves:
* **Classification:** Building a model to predict whether a passenger is "Satisfied" or "Neutral/Dissatisfied" based on various flight and service features.
* **Clustering:** Grouping passengers into clusters based on similar service ratings and travel characteristics to identify different passenger profiles.

## 2. Dataset Source
The dataset was obtained from the **Kaggle** repository.  
**URL:** [Airline Passenger Satisfaction Dataset](https://www.kaggle.com/datasets/teejmahal20/airline-passenger-satisfaction)

## 3. Dataset Description
This dataset contains customer survey data from an airline. It includes:
* **Target Attribute (Class):** `satisfaction` (Categorical: Satisfied, Neutral/Dissatisfied).
* **Number of Instances:** Over 100,000 records.
* **Features:** 25 attributes including:
    * **Demographics:** Gender, Age, Customer Type.
    * **Travel Details:** Type of Travel, Class, Flight Distance.
    * **Service Ratings:** In-flight entertainment, Seat comfort, On-board service, Cleanliness, etc.
